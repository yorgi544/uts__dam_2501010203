````python
import os
import json
import requests
import pandas as pd

# 1. Mengambil data cuaca historis Bali (Denpasar) menggunakan API Open-Meteo
# Koordinat Denpasar: Latitude -8.65, Longitude 115.2167
latitude = -8.65
longitude = 115.2167

url_api = f"https://archive-api.open-meteo.com/v1/archive?latitude={latitude}&longitude={longitude}&start_date=2025-01-01&end_date=2025-01-10&hourly=temperature_2m,relative_humidity_2m,rain&timezone=Asia%2FSingapore"

headers = {
    "User-Agent": "Mahasiswa-Informatika (yorgisimuk@gmail.com)"
}

print("Mengakses API Cuaca...")
response = requests.get(url_api, headers=headers, timeout=30)

if response.status_code == 200:
    # Buat direktori mentah
    os.makedirs("data/raw/api", exist_ok=True)

    # Simpan JSON asli
    raw_json_path = "data/raw/api/cuaca_bali_2025_raw.json"
    data_json = response.json()
    with open(raw_json_path, "w", encoding="utf-8") as f:
        json.dump(data_json, f, indent=4)

    print(f"Data cuaca mentah disimpan di: {raw_json_path}")

    # 2. Parsing JSON menjadi DataFrame terstruktur
    hourly_data = data_json["hourly"]
    df_cuaca = pd.DataFrame({
        "waktu": hourly_data["time"],
        "suhu_celsius": hourly_data["temperature_2m"],
        "kelembaban_persen": hourly_data["relative_humidity_2m"],
        "curah_hujan_mm": hourly_data["rain"]
    })

    # Ekstrak tanggal untuk penggabungan harian
    df_cuaca["tanggal"] = pd.to_datetime(df_cuaca["waktu"]).dt.date

    # Agregasi data cuaca menjadi rata-rata harian
    df_cuaca_harian = df_cuaca.groupby("tanggal").agg({
        "suhu_celsius": "mean",
        "kelembaban_persen": "mean",
        "curah_hujan_mm": "sum"
    }).reset_index()

    # Simpan hasil pemrosesan ke data/processed
    os.makedirs("data/processed", exist_ok=True)
    processed_weather_path = "data/processed/cuaca_bali_processed.csv"
    df_cuaca_harian.to_csv(processed_weather_path, index=False)

    print(f"Data cuaca olahan disimpan di: {processed_weather_path}")
    display(df_cuaca_harian.head())
else:
    print("Gagal mengambil data dari API Cuaca:", response.status_code)
````

# ⚖️ LAPORAN KEPATUHAN ETIKA, LISENSI DATA, DAN BUKTI ROBOTS.TXT

Bagian ini disusun khusus untuk memenuhi rubrik penilaian terkait aspek legalitas, etika, lisensi, dan bukti kepatuhan terhadap protokol perayapan data (*crawling/scraping*).

---

## 1. Bukti Kepatuhan Protokol `robots.txt`

Sebelum melakukan akuisisi data atau perayapan pada situs resmi **Dinas Pariwisata Provinsi Bali** (`https://disparda.baliprov.go.id/`), kami telah melakukan verifikasi terhadap berkas `robots.txt` resmi mereka di URL:
👉 `https://disparda.baliprov.go.id/robots.txt`

### **Hasil Tangkapan Layar / Teks Protokol:**
```text
User-agent: *
Disallow:
```

### **Analisis Kepatuhan:**
* Aturan `Disallow:` yang dikosongkan menunjukkan bahwa pihak pengelola situs **tidak melarang** agen perayap umum (*crawler/scraper*) untuk mengakses halaman publik mereka.
* Pemindaian otomatis menggunakan pustaka Python `urllib.robotparser` (lihat Cell `D4CARBUzSMtO`) secara resmi mengembalikan status nilai **`True`** (Scraping Diizinkan).
* Proyek ini tetap menerapkan batasan etis berupa jeda waktu (*delay*) selama **2 detik** (`time.sleep(2)`) di setiap permintaan untuk menghindari beban berlebih (*overload*) pada peladen (server) instansi pemerintah.

## 3. Catatan Lisensi Penggunaan Data (*Data Licensing*)

### **A. Data Statistik Wisatawan (Disparda Bali)**
* **Jenis Lisensi:** Publik (*Government Open Data Principles* / Domain Publik).
* **Ketentuan Penggunaan:** Sebagai instansi pemerintah daerah, data yang diterbitkan oleh Dinas Pariwisata Provinsi Bali disediakan secara terbuka untuk umum guna mendukung transparansi informasi publik berdasarkan UU Keterbukaan Informasi Publik (KIP) No. 14/2008. Pengguna bebas merujuk dan memproses data ini selama mencantumkan kredit/atribusi ke sumber asli (*Disparda Bali*).

### **B. Data Cuaca Historis (Open-Meteo API)**
* **Jenis Lisensi:** [Creative Commons Attribution 4.0 International (CC BY 4.0)](https://creativecommons.org/licenses/by/4.0/).
* **Ketentuan Penggunaan:** Bebas untuk dibagikan (*share*) dan diadaptasi (*remix/transform*) untuk tujuan apa pun, bahkan secara komersial, dengan syarat wajib mencantumkan atribusi yang sesuai ke platform **Open-Meteo**.

````python
import os
import zipfile
from google.colab import files

zip_name = "proyek_uts_data_acquisition.zip"

# Target file dan folder yang wajib dikumpulkan
targets = ["data", "docs", "notebooks", "README.md"]

print("Memulai pembuatan arsip ZIP proyek...")

try:
    with zipfile.ZipFile(zip_name, 'w', zipfile.ZIP_DEFLATED) as zipf:
        for target in targets:
            if os.path.exists(target):
                if os.path.isdir(target):
                    for root, dirs, files_list in os.walk(target):
                        for file in files_list:
                            file_path = os.path.join(root, file)
                            # Masukkan ke zip dengan path relatif yang rapi
                            zipf.write(file_path, os.path.relpath(file_path, "."))
                else:
                    zipf.write(target, target)
                print(f"✓ Berhasil menambahkan ke ZIP: {target}")
            else:
                print(f"⚠️ Target {target} tidak ditemukan, dilewati.")

    print(f"\n✓ Arsip {zip_name} berhasil dibuat!")
    print("Mencoba memulai pengunduhan otomatis...")
    files.download(zip_name)
except Exception as e:
    print(f"❌ Terjadi kesalahan: {e}")
````
